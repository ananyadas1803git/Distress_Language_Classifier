# 02. Baseline Models: TF-IDF + Logistic Regression & Linear SVM

### Algorithms:
- **Algorithm 1:** TF-IDF + Logistic Regression
- **Algorithm 2:** TF-IDF + Linear SVM (with CalibratedClassifierCV)

```python
import yaml
import pandas as pd
from src.features.tfidf import TFIDFExtractor
from src.models.baseline_models import LogisticRegressionModel, LinearSVMModel
from src.evaluation.metrics import evaluate_all_models

with open('../configs/config.yaml', 'r') as f:
    config = yaml.safe_load(f)

train_df = pd.read_csv('../data/processed/train.csv')
test_df = pd.read_csv('../data/processed/test.csv')

tfidf = TFIDFExtractor(config)
X_train = tfidf.fit_transform(train_df['cleaned_text'])
X_test = tfidf.transform(test_df['cleaned_text'])

lr = LogisticRegressionModel(config).fit(X_train, train_df['label'])
svm = LinearSVMModel(config, calibrate=True).fit(X_train, train_df['label'])

results = {
    'TF-IDF + Logistic Regression': {'y_prob': lr.predict_proba(X_test)[:, 1]},
    'TF-IDF + Linear SVM': {'y_prob': svm.predict_proba(X_test)[:, 1]}
}
evaluate_all_models(results, test_df['label'].values)
```
