# 03. Deep Learning Recurrent Models: LSTM & BiLSTM

### Algorithms:
- **Algorithm 3:** LSTM (Unidirectional)
- **Algorithm 4:** BiLSTM (Bidirectional)

```python
import yaml
import pandas as pd
import torch
from torch.utils.data import DataLoader
from src.data.dataset import Vocabulary, TextClassificationDataset
from src.models.recurrent_models import build_recurrent_model
from src.models.trainer import RecurrentModelTrainer

with open('../configs/config.yaml', 'r') as f:
    config = yaml.safe_load(f)

train_df = pd.read_csv('../data/processed/train.csv')
val_df = pd.read_csv('../data/processed/val.csv')
test_df = pd.read_csv('../data/processed/test.csv')

vocab = Vocabulary()
vocab.build_vocab(train_df['cleaned_text'].tolist(), max_vocab_size=15000)

train_ds = TextClassificationDataset(train_df['cleaned_text'].tolist(), train_df['label'].tolist(), vocab)
val_ds = TextClassificationDataset(val_df['cleaned_text'].tolist(), val_df['label'].tolist(), vocab)
test_ds = TextClassificationDataset(test_df['cleaned_text'].tolist(), test_df['label'].tolist(), vocab)

train_loader = DataLoader(train_ds, batch_size=64, shuffle=True)
val_loader = DataLoader(val_ds, batch_size=64, shuffle=False)
test_loader = DataLoader(test_ds, batch_size=64, shuffle=False)

bilstm_model = build_recurrent_model('bilstm', vocab_size=len(vocab), config=config)
trainer = RecurrentModelTrainer(bilstm_model)
trainer.fit(train_loader, val_loader, epochs=10, checkpoint_path='../saved_models/neural/best_bilstm.pt')
```
